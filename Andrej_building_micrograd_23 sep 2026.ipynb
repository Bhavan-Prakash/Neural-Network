### Derivatives

A **derivative** tells us how much one quantity changes when another quantity changes by a very small amount.

In simple words:

> **A derivative tells us how sensitive `y` is to a change in `x`.**

---

## 1. Simple Example

Consider the function:

$$
y = 2x
$$

If `x` increases by 1, `y` increases by 2.

| x | y |
|---|---|
| 1 | 2 |
| 2 | 4 |
| 3 | 6 |
| 4 | 8 |

Therefore, the rate of change of `y` with respect to `x` is:

$$
\frac{dy}{dx} = 2
$$

We can read this as:

> "How much does `y` change when `x` changes?"

The answer is **2**.

---

## 2. What Information Does a Derivative Give Us?

A derivative gives us information about:

### Direction

If the derivative is **positive**:

$$
\frac{dy}{dx} > 0
$$

then increasing `x` tends to increase `y`.

If the derivative is **negative**:

$$
\frac{dy}{dx} < 0
$$

then increasing `x` tends to decrease `y`.

If the derivative is **zero**:

$$
\frac{dy}{dx} = 0
$$

then a small change in `x` does not change `y` at that point.

### Magnitude

The size of the derivative tells us how strongly `y` responds to a change in `x`.

For example:

$$
\frac{dy}{dx} = 5
$$

means that a small increase in `x` produces approximately 5 times that change in `y`.

---

## 3. Derivative as Slope

Another way to think about a derivative is:

> **The derivative is the slope of a function at a particular point.**

For a straight line, the slope is constant.

For example:

$$
y = 2x
$$

has a constant slope of `2`.

But for a curved function such as:

$$
y = x^2
$$

the slope changes depending on where we are on the curve.

The derivative tells us the slope **at a particular point**.

---

## 4. Example: $y = x^2$

Consider:

$$
y = x^2
$$

Its derivative is:

$$
\frac{dy}{dx} = 2x
$$

So at:

$$
x = 3
$$

the derivative is:

$$
\frac{dy}{dx} = 2(3) = 6
$$

This means:

> At `x = 3`, the function has a slope of `6`.

So if we make a very small change to `x`, `y` will change by approximately `6` times that small change.

---

## 5. Why Are Derivatives Important for Neural Networks?

This is the most important part for understanding **micrograd**.

A neural network has many **weights and parameters**.

We calculate a **loss** that tells us how wrong the network's prediction is.

We then want to know:

> **How does changing each weight affect the loss?**

This is exactly what derivatives can tell us.

For example:

$$
\frac{dL}{dw}
$$

where:

- $L$ = loss
- $w$ = a weight

We can read this as:

> **"How does the loss change when I change the weight?"**

If:

$$
\frac{dL}{dw} > 0
$$

increasing the weight tends to increase the loss.

If:

$$
\frac{dL}{dw} < 0
$$

increasing the weight tends to decrease the loss.

This information allows us to adjust the weights in a direction that reduces the loss.

---

## 6. Connection to Gradient Descent

The basic idea of gradient descent is:

$$
w_{\text{new}} = w_{\text{old}} - \eta \frac{dL}{dw}
$$

where:

- $w$ = weight
- $L$ = loss
- $\eta$ = learning rate
- $\frac{dL}{dw}$ = derivative of the loss with respect to the weight

The derivative tells us **which direction to move**.

The learning rate tells us **how big of a step to take**.

Therefore:

> **Derivatives provide the information needed to update the weights and reduce the loss.**

---

## 7. The Main Idea to Remember

Whenever you see:

$$
\frac{dy}{dx}
$$

think:

> **"How does `y` respond when I slightly change `x`?"**

And when working with neural networks:

$$
\frac{dL}{dw}
$$

think:

> **"How does the loss respond when I slightly change this weight?"**

This idea is fundamental to **backpropagation, gradient descent, and neural network training**.

# What Does "Gradient" Mean?

The word **gradient** can sound complicated, but the basic idea is actually quite simple.

> **A gradient tells us how much the output changes when we slightly change the inputs.**

In simple words:

> **Gradient = direction + amount of change**

---

## 1. Start with a Derivative

For a function with one input:

$$
y = f(x)
$$

we can calculate its derivative:

$$
\frac{dy}{dx}
$$

The derivative tells us:

> **How much does `y` change when `x` changes slightly?**

For example:

$$
y = x^2
$$

The derivative is:

$$
\frac{dy}{dx} = 2x
$$

At:

$$
x = 3
$$

the derivative is:

$$
\frac{dy}{dx} = 6
$$

So around `x = 3`, a small increase in `x` causes `y` to increase by approximately **6 times that small change**.

---

## 2. Then What Is a Gradient?

A **gradient** is essentially the collection of derivatives when we have **multiple inputs**.

Suppose:

$$
L = f(w_1, w_2, w_3)
$$

Here we have three inputs:

- $w_1$
- $w_2$
- $w_3$

We can calculate how the loss changes with respect to each one:

$$
\frac{\partial L}{\partial w_1}
$$

$$
\frac{\partial L}{\partial w_2}
$$

$$
\frac{\partial L}{\partial w_3}
$$

Together, these derivatives form the **gradient**:

$$
\nabla L =
\begin{bmatrix}
\frac{\partial L}{\partial w_1} \\
\frac{\partial L}{\partial w_2} \\
\frac{\partial L}{\partial w_3}
\end{bmatrix}
$$

So you can think of:

> **Derivative → sensitivity to one variable**

> **Gradient → sensitivities to many variables**

---

## 3. Example

Suppose a neural network has two weights:

$$
w_1 = 2
$$

$$
w_2 = 5
$$

And the loss is:

$$
L = f(w_1,w_2)
$$

After calculating the derivatives, suppose we get:

$$
\frac{\partial L}{\partial w_1} = 3
$$

and:

$$
\frac{\partial L}{\partial w_2} = -2
$$

Then the gradient is:

$$
\nabla L =
\begin{bmatrix}
3 \\
-2
\end{bmatrix}
$$

This tells us:

### For $w_1$

$$
\frac{\partial L}{\partial w_1}=3
$$

Increasing $w_1$ tends to **increase the loss**.

### For $w_2$

$$
\frac{\partial L}{\partial w_2}=-2
$$

Increasing $w_2$ tends to **decrease the loss**.

So the gradient gives us information about **how every parameter affects the loss**.

---

## 4. Why Is This Important for Neural Networks?

A neural network has potentially millions or billions of parameters.

For example:

```text
weight 1
weight 2
weight 3
weight 4
...
weight 1,000,000
```

We calculate how the loss changes with respect to each weight:

$$
\frac{\partial L}{\partial w_1},
\frac{\partial L}{\partial w_2},
\frac{\partial L}{\partial w_3},
\ldots
$$

All of these derivatives together form the **gradient**.

The gradient tells the neural network:

> **"If you change each weight, what happens to the loss?"**

---

## 5. Gradient and Gradient Descent

This is where the gradient becomes useful.

Suppose:

$$
\frac{\partial L}{\partial w}=5
$$

The positive gradient tells us:

```text
increase weight
      ↓
loss tends to increase
```

Therefore, to reduce the loss, we want to move the weight in the opposite direction.

If:

$$
\frac{\partial L}{\partial w}=-5
$$

then:

```text
increase weight
      ↓
loss tends to decrease
```

So we want to increase the weight.

This is the basic idea behind **gradient descent**.

The update rule is:

$$
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{\partial L}{\partial w}
$$

where:

- $w$ = weight
- $L$ = loss
- $\eta$ = learning rate
- $\frac{\partial L}{\partial w}$ = gradient for that weight

---

## 6. An Easy Mental Model

Imagine you are standing on a mountain.

The **loss** is the height of the mountain.

You want to get to the bottom.

The gradient tells you:

> **Which direction is uphill, and how steep it is.**

Since you want to go downhill, you move in the **opposite direction of the gradient**.

```text
             High Loss
                /\
               /  \
              /    \
             /      \
            /   YOU  \
           /     ↓    \
          /      ↓     \
         /______↓_______\
             Low Loss
```

---

## 7. Gradient vs Derivative

For now, remember this distinction:

| Concept | Meaning |
|---|---|
| Derivative | How one output changes with respect to one input |
| Gradient | Collection of derivatives with respect to multiple inputs |
| Gradient descent | Using the gradient to move parameters toward lower loss |

For example:

$$
\frac{dL}{dw}
$$

is a **derivative** of the loss with respect to one weight.

But:

$$
\nabla L
$$

represents the **gradient**, containing the derivatives for many parameters.

---

# ⭐ Most Important Thing to Remember

When you hear:

> **"Calculate the gradient."**

Think:

> **"Calculate how the loss changes with respect to every parameter."**

For a neural network:

```text
Weights
   ↓
Forward pass
   ↓
Prediction
   ↓
Loss
   ↓
Backpropagation
   ↓
Gradients
   ↓
Update weights
   ↓
Lower loss
```

So the big picture is:

$$
\boxed{\text{Gradient = information about how changing the parameters changes the loss}}
$$

And this is exactly why gradients are so important in **micrograd, backpropagation, PyTorch, and neural networks**.

# Understanding `self` in Python Classes

One of the most important concepts to understand before going deeper into **micrograd** is `self`.

Consider this class:

    class Value:

        def __init__(self, data):
            self.data = data

        def __repr__(self):
            return f"value of data is : {self.data}"


    data = Value(10)
    print(data)

---

## 1. What is `self`?

`self` refers to the **current object**.

For example:

    data = Value(10)

Here, Python creates an object of the `Value` class.

We can imagine it like this:

    data
     ↓
    ┌─────────────────┐
    │   Value object  │
    │                 │
    │   data = 10     │
    └─────────────────┘

Inside the class, `self` refers to this particular object.

So:

    self.data = data

means:

> "Store the value of `data` inside the current object."

---

## 2. Why do we need `self`?

Suppose we create two objects:

    a = Value(10)
    b = Value(20)

We want each object to remember its own value.

    a
     ↓
    ┌──────────────┐
    │ data = 10    │
    └──────────────┘


    b
     ↓
    ┌──────────────┐
    │ data = 20    │
    └──────────────┘

This is possible because of:

    self.data = data

For object `a`, `self` refers to `a`.

For object `b`, `self` refers to `b`.

So conceptually:

    a.data = 10
    b.data = 20

---

## 3. What happens when we call `Value(10)`?

When we write:

    data = Value(10)

Python creates a new `Value` object and calls its `__init__` method.

Conceptually, we can think of it as:

    __init__(self, 10)

where:

    self → the newly created Value object
    data → 10

Then this line:

    self.data = data

means:

> Put `10` inside the `data` attribute of this object.

We can imagine:

    self
     ↓
    ┌─────────────────┐
    │ data = 10       │
    └─────────────────┘

---

## 4. The Most Important Distinction

Look at this line:

    self.data = data

There are **two different `data`s** here.

### `data` on the right

    self.data = data
                 ↑

This is the **parameter** passed into the function.

For example:

    Value(10)

means that inside `__init__`:

    data = 10

### `self.data` on the left

    self.data = data
    ↑

This is an **attribute belonging to the object**.

Therefore:

    self.data = data

means:

> Take the value from the parameter `data` and store it as `data` inside this particular object.

---

## 5. Another Way to Understand It

Suppose:

    a = Value(10)

Inside `__init__`:

    self.data = data

we can think of it as:

    a.data = 10

Now suppose:

    b = Value(20)

The exact same line:

    self.data = data

can now be thought of as:

    b.data = 20

The reason the same line works for different objects is because `self` refers to the **current object**.

---

## 6. Why is `self` used in `__repr__`?

We have:

    def __repr__(self):
        return f"value of data is : {self.data}"

Again, `self` means:

> "The object on which this method is being called."

When we write:

    print(data)

Python uses the object's `__repr__()` method.

Conceptually:

    data.__repr__()

Therefore, inside `__repr__`:

    self

refers to the `data` object.

So:

    self.data

means:

> The `data` attribute belonging to the current object.

which is:

    10

---

## 7. The Easiest Way to Remember `self`

Whenever you see:

    self.something

think:

> **"something belonging to THIS object."**

For example:

    self.data

means:

> "The `data` belonging to this object."

And:

    self.grad

means:

> "The `grad` belonging to this object."

This becomes very important in **micrograd** because a `Value` object will eventually contain things like:

    self.data
    self.grad
    self._prev
    self._op

Each `Value` object will have its **own** data, gradient, previous nodes, and operation.

---

## 8. Example with Two Objects

Consider:

    class Value:

        def __init__(self, data):
            self.data = data


    a = Value(10)
    b = Value(20)

Think of the objects like this:

    a
     ↓
    ┌─────────────────┐
    │ data = 10       │
    └─────────────────┘


    b
     ↓
    ┌─────────────────┐
    │ data = 20       │
    └─────────────────┘

When this runs:

    a = Value(10)

the line:

    self.data = data

can be thought of as:

    a.data = 10

When this runs:

    b = Value(20)

the same line:

    self.data = data

can be thought of as:

    b.data = 20

So:

> `self` changes depending on the current object.

---

## 9. `self` Is Not a Special Keyword

An important detail:

`self` is a **convention** in Python.

You could technically write:

    class Value:

        def __init__(anything, data):
            anything.data = data

But we don't do this.

Python programmers use:

    self

because it clearly communicates:

> "This is the current object."

So always use `self`.

---

# ⭐ Key Takeaway

The most important thing to remember is:

    self = the current object

Therefore:

    self.data

means:

    data belonging to the current object

And:

    self.grad

means:

    gradient belonging to the current object

Whenever you see `self`, ask yourself:

> **"Which object is currently using this function?"**

That object is `self`.

---

````python
class value:

  def __init__(self, data, _children=(), _op=''):
    self.data = data
    self._prev = set(_children)
    self._op = _op

  def __repr__(self):
    return f"value of data is : {self.data}"


  def __add__(self, other):
    out = value(self.data + other.data, (self, other), '+') # here also we are calling the constructor and passing self.data + other.data to data and (self, other) to _children tuple
    return out

  def __mul__(self, other):
    out = value(self.data * other.data, (self, other), '*')
    return out


data = value(10) + value(20) # it basically mean value(10).__add__(value(20))
data = value(10) * value(20)
data = value(10) * value(20) + value(200)
print(data)
print(data._prev)
data._op
````

````python
from graphviz import Digraph

def trace(root):
  # builds a set of all nodes and edges in a graph
  nodes, edges = set(), set()
  def build(v):
    if v not in nodes:
      nodes.add(v)
      for child in v._prev:
        edges.add((child, v))
        build(child)
  build(root)
  return nodes, edges

def draw_dot(root):
  dot = Digraph(format='svg', graph_attr={'rankdir': 'LR'}) # LR = left to right

  nodes, edges = trace(root)
  for n in nodes:
    uid = str(id(n))
    # for any value in the graph, create a rectangular ('record') node for it
    dot.node(name = uid, label = "{ %s | data %.4f}" % (n.label, n.data), shape='record')
    if n._op:
      # if this value is a result of some operation, create an op node for it
      dot.node(name = uid + n._op, label = n._op)
      # and connect this node to it
      dot.edge(uid + n._op, uid)

  for n1, n2 in edges:
    # connect n1 to the op node of n2
    dot.edge(str(id(n1)), str(id(n2)) + n2._op)

  return dot
````

````python
from graphviz import Digraph

def trace(root):
  # builds a set of all nodes and edges in a graph
  nodes, edges = set(), set()
  def build(v):
    if v not in nodes:
      nodes.add(v)
      for child in v._prev:
        edges.add((child, v))
        build(child)
  build(root)
  return nodes, edges

def draw_dot(root):
  dot = Digraph(format='svg', graph_attr={'rankdir': 'LR'}) # LR = left to right

  nodes, edges = trace(root)
  for n in nodes:
    uid = str(id(n))
    # for any value in the graph, create a rectangular ('record') node for it
    dot.node(name = uid, label = "{ %s | data %.4f | grad %.4f}" % (n.label, n.data, n.grad), shape='record')
    if n._op:
      # if this value is a result of some operation, create an op node for it
      dot.node(name = uid + n._op, label = n._op)
      # and connect this node to it
      dot.edge(uid + n._op, uid)

  for n1, n2 in edges:
    # connect n1 to the op node of n2
    dot.edge(str(id(n1)), str(id(n2)) + n2._op)

  return dot
````

# Understanding the Chain Rule

The **chain rule** is one of the most important concepts for understanding **backpropagation** and **micrograd**.

The basic idea is:

> **If one variable affects another variable, which then affects a third variable, we can calculate the total effect by multiplying the derivatives along the chain.**

---

# 1. What Problem Does the Chain Rule Solve?

Suppose we have:

**c → d → L**

This means:

- `c` affects `d`
- `d` affects `L`
- therefore, `c` indirectly affects `L`

Suppose we want to find:

$$
\frac{dL}{dc}
$$

But we already know:

$$
\frac{dL}{dd}
$$

and:

$$
\frac{dd}{dc}
$$

The **chain rule** tells us how to combine them:

$$
\boxed{
\frac{dL}{dc}
=
\frac{dL}{dd}
\times
\frac{dd}{dc}
}
$$

---

# 2. Why Does Multiplication Make Sense?

Imagine:

**c changes → d changes → L changes**

Suppose a change in `c` causes `d` to change by **2 times as much**.

And suppose a change in `d` causes `L` to change by **3 times as much**.

Then a change in `c` ultimately causes `L` to change by:

$$
2 \times 3 = 6
$$

Therefore:

$$
\frac{dL}{dc}
=
\frac{dL}{dd}
\times
\frac{dd}{dc}
$$

The effects multiply as they pass through the chain.

---

# 3. A Simple Example

Suppose:

$$
d = 2c
$$

and:

$$
L = 3d
$$

So the chain is:

**c → d → L**

Let's say:

$$
c = 5
$$

Then:

$$
d = 2(5) = 10
$$

and:

$$
L = 3(10) = 30
$$

Now let's calculate the derivatives.

Since:

$$
d = 2c
$$

we have:

$$
\frac{dd}{dc}=2
$$

And since:

$$
L = 3d
$$

we have:

$$
\frac{dL}{dd}=3
$$

Using the chain rule:

$$
\frac{dL}{dc}
=
\frac{dL}{dd}
\times
\frac{dd}{dc}
$$

Substitute the values:

$$
\frac{dL}{dc}
=
3 \times 2
$$

Therefore:

$$
\boxed{\frac{dL}{dc}=6}
$$

---

# 4. Think of It as a Chain

This is why it is called the **chain rule**.

We have:

**c → d → L**

We know:

> How does `d` respond to `c`?

This is:

$$
\frac{dd}{dc}
$$

And:

> How does `L` respond to `d`?

This is:

$$
\frac{dL}{dd}
$$

We multiply them:

$$
\frac{dL}{dd}
\times
\frac{dd}{dc}
$$

The `d` in the middle connects the two relationships:

$$
\frac{dL}{\cancel{dd}}
\times
\frac{\cancel{dd}}{dc}
=
\frac{dL}{dc}
$$

This cancellation is a useful way to remember the chain rule.

---

---

# 6. Where Does `dd/dc = 1` Come From?

Suppose:

$$
d = c + e
$$

Then:

$$
\frac{dd}{dc}=1
$$

Why?

Because if we change `c` by 1, `d` also changes by 1.

For example:

**Before:**

- `c = 10`
- `e = 4`
- `d = c + e = 14`

Now increase `c` by 1:

**After:**

- `c = 11`
- `e = 4`
- `d = c + e = 15`

Therefore:

- `c` changed by `+1`
- `d` changed by `+1`

So:

$$
\frac{dd}{dc}
=
\frac{1}{1}
=
1
$$

---

# 7. What About `dd/de = 1`?

We also have:

$$
d = c + e
$$

So changing `e` also changes `d`.

Therefore:

$$
\frac{dd}{de}=1
$$

For example:

**Before:**

- `c = 10`
- `e = 4`
- `d = 14`

If we increase `e`:

**After:**

- `c = 10`
- `e = 5`
- `d = 15`

Again:

- `e` changed by `+1`
- `d` changed by `+1`

Therefore:

$$
\frac{dd}{de}=1
$$

---

**e → d → L**

We already know:

$$
\frac{dL}{dd}=-2
$$

\frac{dL}{dd}

\boxed{\frac{dL}{de}=-2}
$$

If we already know:

$$
\frac{dL}{df}=4
$$

then the effect of `f` on `L` is already known.

If there are no intermediate operations between `f` and `L`, we don't need the chain rule for this particular calculation.

We simply have:

$$
\boxed{\frac{dL}{df}=4}
$$

# 10. The Big Idea Behind Backpropagation

This is the most important part for **micrograd**.

Imagine:

**c → d → L**

During the **forward pass**, information flows forward:

**c → d → L**

We calculate the values.

**L → d → c**

We calculate gradients.

$$

Forward:

**c → d → L**

This is where we calculate the values.

Backward:

**L → d → c**

This is where we calculate the gradients.

Suppose:

$$
\frac{dL}{dd}=-2
$$

That gradient is flowing backward from `L` toward `c`.

At the connection between `d` and `c`, we have:

$$
\frac{dd}{dc}=1
$$

So:

$$

$$

Therefore:

$$
\frac{dL}{dc}=-2
$$

This is the basic mathematical idea behind **backpropagation**.

# 12. General Chain Rule

Whenever we have:

**x → y → z**

and we want to find:

$$
\frac{dz}{dx}
$$

we multiply the derivatives along the path:

$$
\boxed{
\frac{dz}{dx}
=
\frac{dz}{dy}
\times
\frac{dy}{dx}
}
$$

For a longer chain:

**x → a → b → c → L**

we multiply every derivative along the path:

$$
\frac{dL}{dx}
=
\frac{dL}{dc}
\times
\frac{dc}{db}
\times
\frac{db}{da}
\times
\frac{da}{dx}
$$

---

**e ──┘**

- `d` was created from `c` and `e`
- `L` was created from `d`
- each operation has a derivative

During backpropagation, micrograd starts from the loss and works backward through this graph.

At every step, it applies the **chain rule**.

For example:

$$
\frac{dL}{dc}
=
\frac{dL}{dd}
\times
\frac{dd}{dc}
$$

This is why the earlier concepts are connected:

**Computational Graph**

↓

**Chain Rule**

↓

**Backpropagation**

↓

**Gradients**

↓

**Gradient Descent**

# What Is a Forward Pass?

A **forward pass** is the process of taking the input of a neural network and passing it **forward through all the operations** until we get the final output or prediction.

In simple words:

> **Forward pass = input → calculations → prediction → loss**

---

$$

$$

- $x$ = input
- $w$ = weight
- $b$ = bias
- $y$ = output

$$

# 4. Forward Pass vs Backward Pass

This distinction is extremely important for understanding **micrograd**.

## Forward Pass

Information moves **forward**:

    Input → Operations → Prediction → Loss

We calculate the values.

## Backward Pass

Gradients move **backward**:

    Loss → Operations → Parameters

We calculate how much each parameter affected the loss.

So the overall process is:

    FORWARD PASS

    Input
      ↓
    Operations
      ↓
    Prediction
      ↓
    Loss
      ↓
    BACKWARD PASS
      ↓
    Gradients
      ↓
    Update Parameters

we have laid out an mathematical expression and now we're trying to go backward through that expression. going backward through that function means we never want to call a ._backward for any node before we done everything after it. so this ordering of graphs can be achieved using something called topological sort


A topological sort is basically the laying out of a graph so that all the edges only go from left to right so all the edges go one way ( directed graph google it for images )


````python
# below is the topology code

topo = []
visited = set()
def build_topo(v):
  if v not in visited:
    visited.add(v)
    for child in v._prev:
      build_topo(child)
    topo.append(v)

build_topo(f)
topo
````

````python
# now we will add the above topological logic to calculate the gradient of the children in a backward() so that we dont have to
# call _backward on everynode. we will just call that backward func and every gradient will be calculatyed


class value:

  def __init__(self, data, _children=(), _op='', label = ''):
    self.data = data
    self._prev = set(_children)
    self._op = _op
    self.label = label
    self.grad = 0.0
    self._backward = lambda : None #so by default does nothing, this is an empty function, this will be the case for leaf node as theres no previous child

  def __repr__(self):
    return f"value of data is : {self.data}"


  def __add__(self, other):
    out = value(self.data + other.data, (self, other), '+') # here also we are calling the constructor and passing self.data + other.data to data and (self, other) to _children tuple

    def _backward():
      #here we actually want to have the gradien of self and the other and here we will define the function _backward
      #so in an addition the local derivative of an input with the other is just 1.0
      self.grad = 1.0 * out.grad # this is what happen in the addition operations the out.grad gets copied to the children
      other.grad = 1.0 * out.grad # we are just applying chain rules

    out._backward = _backward
    #we used _backward not _backward() because we want to store the _backward not call the function as the func. returns none
    return out

  def __mul__(self, other):
    out = value(self.data * other.data, (self, other), '*')

    def _backward():
      self.grad = other.data * out.grad
      other.grad = self.data * out.grad

    out._backward = _backward
    return out

  def backward(self):

    topo = []
    visited = set()
    def build_topo(v):
      if v not in visited:
        visited.add(v)
        for node in v._prev:
          build_topo(node)
        topo.append(v)

    build_topo(self)

    self.grad = 1.0

    for node in reversed(topo):
      node._backward()
````

we have bug as in if you use two same variable to calculate a target var we get worong differention of the chilren nodes due to problem of overriding

to solve this chain rule something called multivariable case google it, the solution there is basically that we accumulate the gradients i.e plus/add the gradients

so in_backward() of every operation we will just do += instead of = this will be okay as we start the value form 0.0

````python
class value:

  def __init__(self, data, _children=(), _op='', label = ''):
    self.data = data
    self._prev = set(_children)
    self._op = _op
    self.label = label
    self.grad = 0.0
    self._backward = lambda : None #so by default does nothing, this is an empty function, this will be the case for leaf node as theres no previous child

  def __repr__(self):
    return f"value of data is : {self.data}"


  def __add__(self, other):
    out = value(self.data + other.data, (self, other), '+') # here also we are calling the constructor and passing self.data + other.data to data and (self, other) to _children tuple

    def _backward():
      #here we actually want to have the gradien of self and the other and here we will define the function _backward
      #so in an addition the local derivative of an input with the other is just 1.0
      self.grad += 1.0 * out.grad # this is what happen in the addition operations the out.grad gets copied to the children
      other.grad += 1.0 * out.grad # we are just applying chain rules

    out._backward = _backward
    #we used _backward not _backward() because we want to store the _backward not call the function as the func. returns none
    return out

  def __mul__(self, other):
    out = value(self.data * other.data, (self, other), '*')

    def _backward():
      self.grad += other.data * out.grad
      other.grad += self.data * out.grad

    out._backward = _backward
    return out

  def backward(self):

    topo = []
    visited = set()
    def build_topo(v):
      if v not in visited:
        visited.add(v)
        for node in v._prev:
          build_topo(node)
        topo.append(v)

    build_topo(self)

    self.grad = 1.0

    for node in reversed(topo):
      node._backward()
````

now we will do the backpropogations using a deep learning library like pytorch api

in pytorch everything is based around tensors and tensors are just n-dimensional arrays of  scalars

````python
import torch #pytorch

x1 = torch.Tensor([2.0]).double()       ; x1.requires_grad = True #this is similar to what we do a = value(0) i.e creating the value object but here weare not creating value, we're creating tensors
x2 = torch.Tensor([0.0]).double()       ; x2.requires_grad = True #[0.0] means tensor with a single value if it was [0.0,1.0] then it meant tensors with two values
w1 = torch.Tensor([-3.0]).double()       ; w1.requires_grad = True
w2 = torch.Tensor([1.0]).double()       ; w2.requires_grad = True
b = torch.Tensor([6.88137]).double()    ; b.requires_grad = True
n = x1*w1 + x2*w2 + b

o = torch.tanh(n)
print(o.data.item())
o.backward()

print("-----")
print('x2', x2.grad.item())
print('w2', w2.grad.item())
print('x1', x1.grad.item())
print('w1', w1.grad.item())


#these are very basic tensors in actual we would be using

# torch.Tensor([1,2,3], [4,5,6]) i.e, 2x3 matrics

#so above we have done backpropogation using pytorch
````
